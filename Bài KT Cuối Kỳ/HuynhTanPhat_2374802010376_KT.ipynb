### Câu 1 (2 điểm): Cho một ảnh bất kỳ (tên ảnh do sinh viên tự đặt, ví dụ: `my_image.jpg`) và thực hiện các yêu cầu sau:

- Viết chương trình sử dụng median filter để làm mịn ảnh. (0.5 điểm)
- Viết chương trình sử dụng Sobel filter để xác định biên của hình ảnh. (0.5 điểm)
- Đổi màu ảnh bằng cách hoán đổi thứ tự các kênh màu (ví dụ: BGR → GRB hoặc GBR...) và lưu thành tên dạng `[ten_anh]_swapped.jpg`. (0.5 điểm)
- Chuyển ảnh sang không gian màu LAB và tách riêng 3 kênh L, A, B, lưu thành ảnh grayscale tương ứng (`[ten_anh]_L.jpg`, `[ten_anh]_A.jpg`, `[ten_anh]_B.jpg`). (0.5 điểm)


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import imageio.v2 as iio
import scipy.ndimage as sn
from skimage import filters, feature

img = iio.imread('./image1.jpg', mode='F')
img1 = iio.imread('./image1.jpg').astype(np.uint8)
image1_median_filter = sn.median_filter(
    img, size=3, footprint=None, output=None, mode='reflect', cval=0.0, origin=0)
plt.imshow(image1_median_filter)
plt.show()

image1_sobel_filter = filters.sobel(image1_median_filter).astype(np.uint8)
plt.imshow(image1_sobel_filter)
plt.show()

image1_swapped = img1[:, :, [2, 1, 0]]
iio.imsave('./image1_swapped.jpg', image1_swapped)
plt.imshow(image1_swapped)
plt.show()

image1_L = img1[:, :, 0]
image1_A = img1[:, :, 1]
image1_B = img1[:, :, 2]

iio.imsave('./image1_L.jpg', image1_L)
iio.imsave('./image1_A.jpg', image1_A)
iio.imsave('./image1_B.jpg', image1_B)

plt.imshow(image1_L, cmap='gray')
plt.show()

plt.imshow(image1_A, cmap='gray')
plt.show()

plt.imshow(image1_B, cmap='gray')
plt.show()
```

### Câu 2 (4 điểm) Viết một chương trình Python sử dụng OpenCV để tạo menu tương tác cho phép người dùng chọn các kỹ thuật lọc và xử lý ảnh nâng cao từ một danh sách, áp dụng đồng thời cho nhiều ảnh.

### Yêu cầu:

1. Menu gồm:

- Gaussian Blur (kernel size ngẫu nhiên lẻ từ 3 đến 11) (0.5 điểm)
- Median Blur (kernel size ngẫu nhiên lẻ từ 3 đến 11) (0.5 điểm)
- Bilateral Filter (tham số ngẫu nhiên) (0.5 điểm)
- Canny Edge Detection (thresholds ngẫu nhiên 50–150) (0.5 điểm)
- Erosion (kernel 2x2 đến 5x5 ngẫu nhiên) (0.5 điểm)
- Dilation (tương tự Erosion) (0.5 điểm)

2. Chương trình xử lý đồng thời 3 ảnh bất kỳ do sinh viên tự chọn (có thể chọn bằng đường dẫn file hoặc nhập tên ảnh tùy ý). (0.5 điểm)

3. Phím tương ứng để kích hoạt các phương pháp xử lý:

- B: Gaussian Blur
- M: Median Blur
- F: Bilateral Filter
- E: Canny Edge
- R: Erosion
- D: Dilation (0.5 điểm)

4. Lưu file kết quả với định dạng: `result_[phương pháp]_[tên ảnh gốc].jpg`  
   Ví dụ: `result_blur_dog.jpg`, `result_canny_image2.jpg` (0.5 điểm)


```python
import cv2
import numpy as np
import matplotlib.pylab as plt

# Đọc ba ảnh
images = [cv2.imread(f'img{i}.jpg', cv2.IMREAD_GRAYSCALE)
          for i in range(1, 4)]

# Menu động
while True:
    print("Chọn phương pháp biến đổi ảnh:")
    print("B: Gaussian Blur")
    print("M: Median Blur")
    print("F: Bilateral Filter")
    print("E: Canny Edge")
    print("R: Erosion")
    print("D: Dilation")
    print("Q: Thoát")

    choice = input("Nhập lựa chọn của bạn: ").strip().upper()

    if choice == 'Q':
        break

    # Biến đổi ảnh dựa trên lựa chọn
    results = []
    for img in images:
        if choice == 'B':
            result = cv2.GaussianBlur(img, ksize=(5, 5), sigmaX=0)
            cv2.imwrite(f'result_gaussian_blur_img{i + 1}.jpg', result)

        elif choice == 'M':
            result = cv2.medianBlur(img, ksize=5)
            cv2.imwrite(f'result_median_blur_img{i + 1}.jpg', result)

        elif choice == 'F':
            result = cv2.bilateralFilter(
                img, d=9, sigmaColor=75, sigmaSpace=75)
            cv2.imwrite(f'result_bilateral_filter_img{i + 1}.jpg', result)

        elif choice == 'E':
            result = cv2.Canny(img, threshold1=100, threshold2=200)
            cv2.imwrite(f'result_canny_img{i + 1}.jpg', result)

        elif choice == 'R':
            kernel = np.ones((5, 5), np.uint8)
            result = cv2.erode(img, kernel, iterations=1)
            cv2.imwrite(f'result_erosion_img{i + 1}.jpg', result)

        elif choice == 'D':
            kernel = np.ones((5, 5), np.uint8)
            result = cv2.dilate(img, kernel, iterations=1)
            cv2.imwrite(f'result_dilation_img{i + 1}.jpg', result)

        else:
            print("Lựa chọn không hợp lệ!")
            continue

        results.append(result)

    # Hiển thị ảnh gốc và ảnh đã biến đổi
    plt.figure(figsize=(15, 10))
    for i in range(3):
        plt.subplot(3, 2, 2 * i + 1)
        plt.imshow(images[i])
        plt.title(f'Ảnh gốc {i + 1}')
        plt.axis('off')

        plt.subplot(3, 2, 2 * i + 2)
        plt.imshow(results[i])
        plt.title(f'Ảnh đã biến đổi {i + 1}')
        plt.axis('off')

    plt.show()

print("Chương trình đã kết thúc.")
```

### Câu 3 (4 điểm) Viết một chương trình Python để xử lý 3 ảnh bất kỳ do sinh viên tự chọn.

- Tăng kích thước ảnh đầu tiên thêm 35 pixel ở mỗi cạnh. (0.5 điểm)
- Xoay ảnh thứ hai 135 độ và lật ngang. (0.5 điểm)
- Tăng kích thước ảnh thứ ba lên 5 lần và làm mịn bằng Gaussian blur kernel 9x9. (1.5 điểm)
- Thay đổi độ sáng và độ tương phản ảnh thứ ba bằng công thức:

$$
I_{out}(x, y) = \alpha \cdot I_{in}(x, y) + \beta
$$

Trong đó:

$$
\alpha \in [0.5, 2.0], \quad \beta \in [-50, 50]
$$

Giá trị đầu ra cần được giới hạn trong khoảng [0, 255] bằng công thức:

$$
I_{out}(x, y) = \text{clip}(I_{out}(x, y), 0, 255)
$$


```python
import cv2
import matplotlib.pylab as plt
import numpy as np

# Đọc ba ảnh
img1 = cv2.imread('./img1.jpg')
img2 = cv2.imread('./img2.jpg')
img3 = cv2.imread('./img3.jpg')

# Tăng kích thướt ảnh
img1_resized = cv2.resize(img1, (img1.shape[1] + 35, img1.shape[0] + 35))
plt.imshow(cv2.cvtColor(img1_resized, cv2.COLOR_BGR2RGB))
plt.show()

# Xoay ảnh và lật ngang
(h, w) = img2.shape[:2]
center = (w // 2, h // 2)
M = cv2.getRotationMatrix2D(center, -135, 1.0)  # Xoay 135 độ
img2_rotated = cv2.warpAffine(img2, M, (w, h))
img2_flipped = cv2.flip(img2_rotated, 1)  # Lật ngang
plt.imshow(cv2.cvtColor(img2_flipped, cv2.COLOR_BGR2RGB))
plt.show()

# Tăng kích thướt ảnh lên 5 lần và làm mịn bằng Gaussian blur kernel 9x9
img3_resized = cv2.resize(img3, (img3.shape[1] * 5, img3.shape[0] * 5))
img3_blurred = cv2.GaussianBlur(img3_resized, (9, 9), 0)
plt.imshow(cv2.cvtColor(img3_blurred, cv2.COLOR_BGR2RGB))
plt.show()

# Áp dụng công thức
alpha = np.random.uniform(0.5, 2.0)  # Hệ số tương phản
beta = np.random.randint(-50, 51)    # Độ sáng
output_img3 = cv2.convertScaleAbs(img3, alpha=alpha, beta=beta)
# Giới hạn giá trị đầu ra trong khoảng [0, 255]
output_img3 = np.clip(output_img3, 0, 255).astype(np.uint8)
plt.imshow(cv2.cvtColor(output_img3, cv2.COLOR_BGR2RGB))
plt.show()
```
