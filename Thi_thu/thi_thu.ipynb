```python
# Câu 1 (2 Điểm): Cho ảnh có tên là a.jpg và thực hiện các yêu cầu:
# - Viết chương trình sử dụng mean filter cho ảnh (0.5 Điểm)
# - Viết chương trình sử dụng filter xác định biên của hình ảnh trên(0.5 Điểm)
# - Viết chương trình đổi màu ảnh từ không gian màu BGR sang một màu ngẫu
# nhiên (RGB) bằng cách thay đổi các kênh màu một cách ngẫu nhiên, sau đó lưu
# hình mới vào file a_random_color.jpg(0.5 Điểm)
# - Chuyển ảnh sang không gian màu HSV và tách riêng kênh Hue, Saturation, Value
# để lưu thành ba ảnh grayscale tương ứng (a_hue.jpg, a_saturation.jpg,
# a_value.jpg) (0.5 Điểm)

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Đọc ảnh
img = cv2.imread('./a.jpg')

# Sử dụng mean filter
mean_filter = np.ones((3, 3)) / 9
mean_filtered_img = cv2.filter2D(img, -1, mean_filter)

# Sử dụng filter xác định biên
sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
sobel_y = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])
sobel_filtered_img = cv2.filter2D(img, -1, sobel_x) + cv2.filter2D(img, -1, sobel_y)

# Chuyển màu sang màu ngẫu nhiên
b_random = np.random.randint(0, 256, img.shape[:2], dtype=np.uint8)
g_random = np.random.randint(0, 256, img.shape[:2], dtype=np.uint8)
r_random = np.random.randint(0, 256, img.shape[:2], dtype=np.uint8)
img_random_color = cv2.merge((b_random, g_random, r_random))
cv2.imwrite('./a_random_color.jpg', img_random_color)

# Chuyển màu sang màu HSV
hsv_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv_img)

# Lưu các kênh Hue, Saturation, Value
cv2.imwrite('./a_hue.jpg', h)
cv2.imwrite('./a_saturation.jpg', s)
cv2.imwrite('./a_value.jpg', v)

# Hiển thị ảnh
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(mean_filtered_img, cv2.COLOR_BGR2RGB))
plt.title('Mean Filtered Image')

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(sobel_filtered_img, cv2.COLOR_BGR2RGB))
plt.title('Sobel Filtered Image')

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(img_random_color, cv2.COLOR_BGR2RGB))
plt.title('Random Color Image')

plt.show()
```

```python
# Câu 2 (4 Điểm): Viết một chương trình Python sử dụng OpenCV để tạo menu động cho
# phép người dùng chọn các phương pháp biến đổi ảnh từ một danh sách mở rộng, áp
# dụng cho nhiều ảnh cùng lúc, và thực hiện các phân tích bổ sung. Các yêu cầu cụ thể:
# 1. Tạo một menu động cho phép người dùng chọn các phương pháp biến đổi ảnh sau:
# - Image inverse transformation(0.5 Điểm)
# - Gamma-Correction (với giá trị gamma ngẫu nhiên từ 0.5 đến 2.0) (0.5 Điểm)
# - Log Transformation (với hệ số nhân ngẫu nhiên từ 1.0 đến 5.0) (0.5 Điểm)
# - Histogram equalization(0.5 Điểm)
# - Contrast Stretching (tùy chỉnh với giá trị min và max ngẫu nhiên từ 0 đến 255) (0.5 Điểm)
# - Adaptive Histogram Equalization (sử dụng CLAHE với ô lưới 8x8) (0.5 Điểm)
# 2. Chương trình phải xử lý đồng thời ba ảnh đầu vào có tên image1.jpg, image2.jpg, và image3.jpg (0.5 Điểm)
# 3. Khi người dùng nhấn các phím sau, chương trình sẽ thực hiện biến đổi tương ứng
# trên cả ba ảnh (0.5 Điểm)
# - Phím I: Image inverse transformation
# - Phím G: Gamma-Correction
# - Phím L: Log Transformation
# - Phím H: Histogram equalization
# - Phím C: Contrast Stretching
# - Phím A: Adaptive Histogram Equalization
# 4. Đối với mỗi biến đổi, lưu kết quả của từng ảnh vào các file riêng biệt với tên định dạng output_[phương pháp]_[số ảnh].jpg (ví dụ: output_inverse_1.jpg, output_gamma_2.jpg).

import cv2
import numpy as np
import matplotlib.pylab as plt
import random

# Hàm biến đổi ảnh
def image_inverse(img):
    return 255 - img

# Hàm gamma correction
def gamma_correction(img, gamma):
    inv_gamma = 1.0 / gamma
    table = np.array([(i / 255.0) ** inv_gamma * 255 for i in range(256)]).astype("uint8")
    return cv2.LUT(img, table)

# Hàm log transformation
def log_transformation(img, c):
    img_log = c * np.log1p(img)
    img_log = np.array(np.clip(img_log, 0, 255), dtype=np.uint8)
    return img_log

# histogram equalization
def histogram_equalization(img):
    return cv2.equalizeHist(img)

# contrast stretching
def contrast_stretching(img, min_val, max_val):
    img_stretched = cv2.normalize(img, None, alpha=min_val, beta=max_val, norm_type=cv2.NORM_MINMAX)
    return img_stretched

# adaptive histogram equalization
def adaptive_histogram_equalization(img):
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    return clahe.apply(img)

# Đọc ba ảnh
images = [cv2.imread(f'image{i}.jpg', cv2.IMREAD_GRAYSCALE) for i in range(1, 4)]

# Menu động
while True:
    print("Chọn phương pháp biến đổi ảnh:")
    print("I: Image inverse transformation")
    print("G: Gamma-Correction")
    print("L: Log Transformation")
    print("H: Histogram equalization")
    print("C: Contrast Stretching")
    print("A: Adaptive Histogram Equalization")
    print("Q: Thoát")

    choice = input("Nhập lựa chọn của bạn: ").strip().upper()

    if choice == 'Q':
        break

    for i, img in enumerate(images):
        if choice == 'I':
            result = image_inverse(img)
            cv2.imwrite(f'output_inverse_{i + 1}.jpg', result)

        elif choice == 'G':
            gamma = random.uniform(0.5, 2.0)
            result = gamma_correction(img, gamma)
            cv2.imwrite(f'output_gamma_{i + 1}.jpg', result)

        elif choice == 'L':
            c = random.uniform(1.0, 5.0)
            result = log_transformation(img, c)
            cv2.imwrite(f'output_log_{i + 1}.jpg', result)

        elif choice == 'H':
            result = histogram_equalization(img)
            cv2.imwrite(f'output_histogram_{i + 1}.jpg', result)

        elif choice == 'C':
            min_val = random.randint(0, 255)
            max_val = random.randint(min_val, 255)
            result = contrast_stretching(img, min_val, max_val)
            cv2.imwrite(f'output_contrast_{i + 1}.jpg', result)

        elif choice == 'A':
            result = adaptive_histogram_equalization(img)
            cv2.imwrite(f'output_adaptive_{i + 1}.jpg', result)

        else:
            print("Lựa chọn không hợp lệ!")
            
        # Hiển thị ảnh gốc và ảnh đã biến đổi
        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        plt.imshow(img, cmap='gray')
        plt.title(f'Original Image {i + 1}')
        plt.axis('off')

        plt.subplot(1, 2, 2)
        plt.imshow(result, cmap='gray')
        plt.title(f'Transformed Image {i + 1}')
        plt.axis('off')

        plt.show()

print("Chương trình đã kết thúc.")
```

```python
# Câu 3 (4 Điểm): Viết một chương trình Python sử dụng OpenCV để xử lý ba ảnh:
# colorful-ripe-tropical-fruits.jpg, quang-ninh.jpg, và pagoda.jpg với các phương pháp
# biến đổi và tiền xử lý nâng cao.
# - Tăng kích thước ảnh colorful-ripe-tropical-fruits.jpg thêm 30 pixel ở cả chiều rộng và chiều cao. (0.5 Điểm)
# - Xoay ảnh quang_ninh.jpg 45 độ theo chiều kim đồng hồ và lật ngang. (0.5 Điểm)
# - Tăng kích thước ảnh pagoda.jpg lên 5 lần và áp dụng Gaussian blur với kernel 7x7 để làm mịn. (0.5 Điểm)
# - Ứng dụng công thức bên dưới cho ảnh pagoda.jpg (1.5 Điểm)

import cv2
import matplotlib.pylab as plt
import numpy as np

# Đọc ba ảnh
img1 = cv2.imread('./colorful-ripe-tropical-fruits.jpg')
img2 = cv2.imread('./quang_ninh.jpg')
img3 = cv2.imread('./pagoda.jpg')

# Tăng kích thước ảnh colorful-ripe-tropical-fruits.jpg thêm 30 pixel ở cả chiều rộng và chiều cao
img1_resized = cv2.resize(img1, (img1.shape[1] + 30, img1.shape[0] + 30))

# Xoay ảnh quang-ninh.jpg 45 độ theo chiều kim đồng hồ và lật ngang
(h, w) = img2.shape[:2]
center = (w // 2, h // 2)
M = cv2.getRotationMatrix2D(center, -45, 1.0)  # Xoay 45 độ
img2_rotated = cv2.warpAffine(img2, M, (w, h))
img2_flipped = cv2.flip(img2_rotated, 1)  # Lật ngang

# Tăng kích thước ảnh pagoda.jpg lên 5 lần và áp dụng Gaussian blur với kernel 7x7
img3_resized = cv2.resize(img3, (img3.shape[1] * 5, img3.shape[0] * 5))
img3_blurred = cv2.GaussianBlur(img3_resized, (7, 7), 0)

# Áp dụng công thức
alpha = np.random.uniform(0.5, 2.0)  # Hệ số tương phản
beta = np.random.randint(-50, 51)    # Độ sáng
output_img3 = cv2.convertScaleAbs(img3_resized, alpha=alpha, beta=beta)

# Giới hạn giá trị đầu ra trong khoảng [0, 255]
output_img3 = np.clip(output_img3, 0, 255).astype(np.uint8)

# Hiển thị kết quả
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img1_resized, cv2.COLOR_BGR2RGB))
plt.title('Resized Colorful Fruits')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(img2_flipped, cv2.COLOR_BGR2RGB))
plt.title('Rotated and Flipped Quang Ninh')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(img3_blurred, cv2.COLOR_BGR2RGB))
plt.title('Resized and Blurred Pagoda')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(output_img3, cv2.COLOR_BGR2RGB))
plt.title('Combined Pagoda Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
